```python
import os
import torch
import numpy as np
import matplotlib.pyplot as plt
from rototransl_env import TrackingEnv
from ddpg_dyn import DDPGAgent as DDPGTranslAgent


# ---------- LOAD TRANSLATION-ONLY AGENT ----------
def load_transl_agent(checkpoint_path_transl, env=None):
    if env is None:
        env = TrackingEnv()

    # stato: x, y, x_target, y_target
    state_dim_trasl = 4
    agent_transl = DDPGTranslAgent(state_dim_trasl, 2)

    ckpt_transl = torch.load(checkpoint_path_transl, map_location="cpu")
    agent_transl.actor.load_state_dict(ckpt_transl["actor_state_dict"])
    agent_transl.eval()
    return agent_transl


# ---------- TEST TRANSLATION-ONLY ----------
def test_transl_agent(
    agent_transl,
    env=None,
    num_episodes=1,
    tolerance_transl=0.02,
):
    if env is None:
        env = TrackingEnv()

    final_distances, ep_attached_counter = [], []
    success_counter, fail_counter = 0, 0

    for ep in range(num_episodes):
        state, _ = env.reset()
        real_state = torch.tensor(state, dtype=torch.float32)
        state = torch.tensor(state, dtype=torch.float32)

        state = state.clone()
        #state[3:5] += torch.normal(0.0, 0.005, size=(2,))
        done, step = False, 0
        trajectory, target_traj = [], []
        total_attached_counter = 0

        while not done:
            step += 1
            trajectory.append(state[:2].numpy())
            target_traj.append(state[3:5].numpy())

            # costruiamo l’osservazione per l’attore di traslazione
            state_pos = torch.cat([state[:2], state[3:5]], dim=0)
            with torch.no_grad():
                action_xy = agent_transl.actor(state_pos)

            # azione rotazione forzata a 0
            action = torch.cat([action_xy, torch.zeros(1)], dim=0).numpy()
            action = np.clip(action, env.action_space.low, env.action_space.high)

            next_state, _, done, truncated, _, _ = env.step(action)
            real_next_state = torch.tensor(next_state, dtype=torch.float32)
            next_state = torch.tensor(next_state, dtype=torch.float32)
    
            next_state = next_state.clone()
            #next_state[3:5] += torch.normal(0.0, 0.005, size=(2,))

            # distanza agente–target
            dist_transl = torch.norm(real_next_state[:2] - real_state[3:5])
            if dist_transl < tolerance_transl:
                total_attached_counter += 1

            done = truncated
            real_state = real_next_state
            state = next_state

        final_distances.append(dist_transl.item())
        ep_attached_counter.append(total_attached_counter)

        success = total_attached_counter > 80
        fail = total_attached_counter < 70
        success_counter += int(success)
        fail_counter += int(fail)

        print(
            f"[Ep {ep}] steps={step}  attached={total_attached_counter} "
            f"final_dist={dist_transl:.4f}  success={success}"
        )

        if ep == 0:
            plot_xy_trajectory(trajectory, target_traj, ep, success)

    env.close()
    print("\n--- STATISTICHE ---")
    print(f"Media distanza finale     : {np.mean(final_distances):.4f}")
    print(f"Media passi attaccati     : {np.mean(ep_attached_counter):.2f}")
    print(f"Successi totali           : {success_counter}/{num_episodes}")
    print(f"Fail totali               : {fail_counter}/{num_episodes}")


# ---------- PLOT ----------
def plot_xy_trajectory(trajectory, target_traj, episode, success):
    traj = np.array(trajectory)
    tgt = np.array(target_traj)

    plt.figure(figsize=(6, 5))
    plt.plot(traj[:, 0], traj[:, 1], "o-", label="Agent")
    plt.plot(tgt[:, 0], tgt[:, 1], "x--", label="Target")
    plt.scatter(traj[0, 0], traj[0, 1], c="green", s=80, label="Start agent")
    plt.scatter(tgt[0, 0], tgt[0, 1], c="yellow", s=80, label="Start target")
    plt.title(f"Agent and target trajectories")
    plt.xlabel("X")
    plt.ylabel("Y")
    plt.axis("equal")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()
    plt.close()


# ---------- MAIN ----------
if __name__ == "__main__":
    ckpt_transl = ("/Users/edoardozappia/Documents/GitHub/RL_IL_AR_for_surgery/Traslazioni-dinamiche/No-noise/ddpg_run_dyn20250416_200754/checkpoint_ep2950.pth")

    agent_transl = load_transl_agent(ckpt_transl)
    test_transl_agent(agent_transl)
```

```python
import os
import torch
import numpy as np
import matplotlib.pyplot as plt
from env_rot import TrackingEnv
from ddpg_rot_dyn import DDPGAgent

def load_agent(checkpoint_path, env=None):
    if env is None:
        env = TrackingEnv()
    state_dim = 2 #env.observation_space.shape[0]
    action_dim = 1 #env.action_space.shape[0]
    agent = DDPGAgent(state_dim, action_dim)

    checkpoint = torch.load(checkpoint_path, map_location=torch.device('cpu'))
    agent.actor.load_state_dict(checkpoint['actor_state_dict'])
    agent.critic.load_state_dict(checkpoint['critic_state_dict'])

    agent.eval()  # Imposta .eval() anche sui moduli interni (se serve, aggiungi)
    return agent

def test_agent_with_random_targets(agent, env=None, num_episodes=1, tolerance=0.01):
    if env is None:
        env = TrackingEnv()

    ep_attached_counter = []
    success_counter = 0
    total_dist = 0


    for ep in range(num_episodes):
        state, _ = env.reset()

        real_state = torch.tensor(state, dtype=torch.float32)
        state = torch.tensor(state, dtype=torch.float32)
        
        state = state.clone()
        state[1:] += torch.normal(mean=0.0, std=0.004, size=(1,), device=state.device)

        done = False
        step = 0
        trajectory = []
        target_trajectory = []
        attached_counter = 0
        total_attached_counter = 0

        while not done:
            trajectory.append(state[0].detach().numpy())
            target_trajectory.append(state[1].detach().numpy())
            with torch.no_grad():
                action = agent.actor(state).detach().numpy()
            next_state, _, done, truncated, _ = env.step(action)

            real_next_state = torch.tensor(next_state, dtype=torch.float32)
            next_state = torch.tensor(next_state, dtype=torch.float32)

            next_state = next_state.clone()
            next_state[1:] += torch.normal(mean=0.0, std=0.004, size=(1,), device=next_state.device)

            dist = torch.norm(real_next_state[0] - real_state[1])

            if dist < tolerance:
                total_attached_counter += 1
                attached_counter += 1
                total_dist += dist
            else:
                attached_counter = 0

            done = truncated

            state = next_state
            real_state = real_next_state
            step += 1

        ep_attached_counter.append(total_attached_counter)

        print(f"[Episode {ep}] Steps: {step}, Total attached counter: {total_attached_counter}")

        if ep == 0:
            trajectory = np.array(trajectory)
            target_trajectory = np.array(target_trajectory)
            plt.figure(figsize=(6, 4))
            plt.plot(trajectory, "o-", label="Agent")
            plt.plot(target_trajectory, "x--", label="Target")
            plt.scatter(0, trajectory[0], color='green', label='Start agent', s=80)
            plt.scatter(0, target_trajectory[0], color='yellow', label='Start target', s=80)
            plt.title(f"Rotation agent and target trajectories")
            plt.xlabel("Steps")
            plt.ylabel("Angles (rad)")
            plt.legend()
            plt.grid(True)
            plt.tight_layout()            
            plt.show()                    

    env.close()
    print(f"\nMedia passi attaccati: {np.mean(ep_attached_counter):.2f}")
    print(f"Media distanza finale: {total_dist / (num_episodes*100):.4f} ")
    #print(f"Numero successi: {success_counter} / {num_episodes}")

if __name__ == "__main__":
    checkpoint = "/Users/edoardozappia/Documents/GitHub/RL_IL_AR_for_surgery/TEST_NOISE/Rotazioni-dinamiche/ddpg_mov_0.01_std_0.004_20250620_222306/checkpoint_ep2471.pth"
    agent = load_agent(checkpoint)
    test_agent_with_random_targets(agent)
```

```python
import os
import torch
import numpy as np
import matplotlib.pyplot as plt
from rototransl_env import TrackingEnv
from ddpg_dyn import DDPGAgent as DDPGTranslAgent
from ddpg_rot_dyn import DDPGAgent as DDPGRotAgent


def load_agents(checkpoint_path_transl, checkpoint_path_rot, env=None):
    if env is None:
        env = TrackingEnv()

    state_dim_trasl = 4  # x, y, x_target, y_target
    state_dim_rot = 2    # theta, theta_target

    agent_transl = DDPGTranslAgent(state_dim_trasl, 2)
    agent_rot = DDPGRotAgent(state_dim_rot, 1)

    ckpt_transl = torch.load(checkpoint_path_transl, map_location=torch.device('cpu'))
    ckpt_rot = torch.load(checkpoint_path_rot, map_location=torch.device('cpu'))

    agent_transl.actor.load_state_dict(ckpt_transl['actor_state_dict'])
    agent_rot.actor.load_state_dict(ckpt_rot['actor_state_dict'])

    agent_transl.eval()
    agent_rot.eval()

    return agent_transl, agent_rot

def test_dual_agents(agent_transl, agent_rot, env=None, num_episodes=1, tolerance_transl=0.02, tolerance_rot=0.01):
    if env is None:
        env = TrackingEnv()

    final_distances = []
    final_rot_errors = []
    ep_attached_counter = []
    ep_attached_counter_transl = []
    ep_attached_counter_rot = []
    success_counter = 0

    for ep in range(num_episodes):
        state, _ = env.reset()
        real_state = torch.tensor(state, dtype=torch.float32)
        state = torch.tensor(state, dtype=torch.float32)

        state = state.clone()
        state[3:5] += torch.normal(mean=0.0, std=0.005, size=(2,), device=state.device)
        state[5:] += torch.normal(mean=0.0, std=0.004, size=(1,), device=state.device)

        done = False
        step = 0
        trajectory = []
        target_traj = []
        theta_traj = []
        theta_target_traj = []
        attached_counter = 0
        total_attached_counter = 0
        total_attached_counter_transl = 0
        total_attached_counter_rot = 0

        while not done:
            step += 1
            trajectory.append(state[:2].detach().numpy())
            target_traj.append(state[3:5].detach().numpy())
            theta_traj.append(state[2].item())
            theta_target_traj.append(state[5].item())

            with torch.no_grad():
                state_pos = torch.cat([state[:2], state[3:5]], dim=0)
                state_rot = torch.cat([state[2:3], state[5:6]], dim=0)
                action_xy = agent_transl.actor(state_pos)
                action_rot = agent_rot.actor(state_rot)

            action = torch.cat([action_xy, action_rot], dim=0).detach().numpy()
            #print("action prima del clipping: ", action)
            action = np.clip(action, env.action_space.low, env.action_space.high)
            #print("action dopo clipping: ", action)

            next_state, _, done, truncated, _, _ = env.step(action)
            real_next_state = torch.tensor(next_state, dtype=torch.float32)
            next_state = torch.tensor(next_state, dtype=torch.float32)

            next_state = next_state.clone()
            next_state[3:5] += torch.normal(mean=0.0, std=0.005, size=(2,), device=next_state.device)
            next_state[5:] += torch.normal(mean=0.0, std=0.004, size=(1,), device=next_state.device)

            dist_transl = torch.norm(real_next_state[:2] - real_state[3:5])
            dist_rot = torch.abs(real_next_state[2] - real_state[5])

            if dist_transl < tolerance_transl:
                total_attached_counter_transl += 1

            if dist_rot < tolerance_rot:
                total_attached_counter_rot += 1

            if dist_transl < tolerance_transl and dist_rot < tolerance_rot:
                total_attached_counter += 1
                attached_counter += 1
            else:
                attached_counter = 0

            done = truncated

            state = next_state
            real_state = real_next_state

        final_distances.append(dist_transl.item())
        final_rot_errors.append(dist_rot.item())
        ep_attached_counter.append(total_attached_counter)
        ep_attached_counter_transl.append(total_attached_counter_transl)
        ep_attached_counter_rot.append(total_attached_counter_rot)

        success = attached_counter > 20
        if success:
            success_counter += 1

        #print(f"[Episode {ep}] Steps: {step}, Total attached counter: {total_attached_counter}, Total transl: {total_attached_counter_transl}, Total rot: {total_attached_counter_rot}, Final distance: {dist_transl:.4f}, Final rot error: {dist_rot:.4f}, Success: {success}")

        if ep == 0:
            plot_trajectory(trajectory, target_traj, theta_traj, theta_target_traj, ep, success)

    env.close()
    print(f"\nMedia distanza finale: {np.mean(final_distances):.4f}")
    print(f"Media errore rotazione finale: {np.mean(final_rot_errors):.4f}")
    print(f"\nMedia passi attaccati: {np.mean(ep_attached_counter):.4f}")
    print(f"\nMedia passi attaccati traslazione: {np.mean(ep_attached_counter_transl):.4f}")
    print(f"\nMedia passi attaccati rotazione: {np.mean(ep_attached_counter_rot):.4f}")
    print(f"Successi totali: {success_counter} / {num_episodes}")

def plot_trajectory(trajectory, target_traj, theta_traj, theta_target_traj, episode, success):
    trajectory = np.array(trajectory)
    target_traj = np.array(target_traj)
    theta_traj = np.array(theta_traj)
    theta_target_traj = np.array(theta_target_traj)

    fig, axs = plt.subplots(1, 2, figsize=(12, 5))

    axs[0].plot(trajectory[:, 0], trajectory[:, 1], marker='o', label="Agent")
    axs[0].plot(target_traj[:, 0], target_traj[:, 1], marker='x', linestyle='--', label="Target")
    axs[0].scatter(trajectory[0, 0], trajectory[0, 1], color='green', label='Start Agent', s=100)
    axs[0].scatter(target_traj[0, 0], target_traj[0, 1], color='yellow', label='Start Target', s=100)
    axs[0].set_title("Translation trajectories")
    axs[0].set_xlabel("X")
    axs[0].set_ylabel("Y")
    axs[0].legend()
    axs[0].grid(True)
    axs[0].axis('equal')

    axs[1].plot(theta_traj, label="Agent", marker='o')
    axs[1].plot(theta_target_traj, label="Target", linestyle='--', marker='x')
    axs[1].scatter(0, theta_traj[0], color='green', label='Start Agent', s=100)
    axs[1].scatter(0, theta_target_traj[0], color='yellow', label='Start Target', s=100)
    axs[1].set_title("Rotation trajectories")
    axs[1].set_xlabel("Steps")
    axs[1].set_ylabel("Angles")
    axs[1].legend()
    axs[1].grid(True)

    plt.suptitle(f"Agent and target trajectories")
    plt.tight_layout()
    plt.show()
    plt.close()

# --- MAIN ---
if __name__ == "__main__":
    ckpt_transl = "/Users/edoardozappia/Documents/GitHub/RL_IL_AR_for_surgery/Traslazioni-dinamiche/No-noise/ddpg_run_dyn20250416_200754/checkpoint_ep2950.pth"
    ckpt_rot = "/Users/edoardozappia/Documents/GitHub/RL_IL_AR_for_surgery/Rotazioni-dinamiche/No-noise/ddpg_mov_0.01_20250509_163508/checkpoint_ep782.pth"
    agent_transl, agent_rot = load_agents(ckpt_transl, ckpt_rot)
    test_dual_agents(agent_transl, agent_rot)
```
